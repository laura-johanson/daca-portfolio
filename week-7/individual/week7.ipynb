1. Määran viitekuupäeva, kui kaua on kliendi viimasest ostust möödas.

```python
today = pd.to_datetime('2025-02-28')
```

2. Leian iga kliendi viimase ostu kuupäeva.

```python
last_purchase = df[df['sale_date'] <= today].groupby('customer_id')['sale_date'].max()
```

3. Arvutan Recency (viitekuupäev − kliendi viimane ost = mitu päeva on viimasest ostust möödas).

```python
recency = (today - last_purchase).dt.days
```

4. Arvutan Frequency (iga kliendi ostude arv).

```python
frequency = df.groupby('customer_id')['sale_id'].count()
```

5. Arvutan Monetary (iga kliendi kogukulutus).

```python
monetary = df.groupby('customer_id')['total_price'].sum()
```

6. Ühendan recency, frequency ja monetary. 

```python
rfm = pd.merge(recency, frequency, on='customer_id')
rfm = pd.merge(rfm, monetary, on='customer_id')

rfm.columns = ['Recency', 'Frequency', 'Monetary']
```

7. Määran Recency skoori (hiljuti ostnud klient saab kõrgema skoori).

```python
rfm['R_score'] = pd.qcut(rfm['Recency'], 5, labels=[5, 4, 3, 2, 1])
```

8. Paljudel klientidel kordub täpselt sama ostude arv, seega kasutan rank()-i. 


```python
frequency_rank = rfm['Frequency'].rank(method='first')
```

9. Määran Frequency skoori.

```python
rfm['F_score'] = pd.qcut(frequency_rank, 5, labels=[1, 2, 3, 4, 5])
```

10. Määran Monetary skoori.

```python
rfm['M_score'] = pd.qcut(rfm['Monetary'], 5, labels=[1, 2, 3, 4, 5])
```

11. Arvutan RFM koguskoori.

```python
rfm['RFM_Score'] = (
    rfm['R_score'].astype(int)
    + rfm['F_score'].astype(int)
    + rfm['M_score'].astype(int)
)
```

12. Kontrollin, millised skoorid tekkisid.

```python
print(rfm['RFM_Score'].value_counts().sort_index())
```

13. Loon segmendid (13-15 = VIP Champions, 10-12 = Loyal, 7-9 = Potential, 4-6 = At Risk, 3 = Lost). 

```python
def segment(score):
    if score >= 13:
        return 'VIP Champions'
    elif score >= 10:
        return 'Loyal'
    elif score >= 7:
        return 'Potential'
    elif score >= 4:
        return 'At Risk'
    else:
        return 'Lost'

rfm['Segment'] = rfm['RFM_Score'].apply(segment)
```

14. Segmentide kokkuvõte.

```python
print(rfm['Segment'].value_counts().sort_index())
```

1. Impordin Plotly.

```python
import plotly.express as px
```

2. Diagrammi 1 - segmentide jaotus.

```python
segment_counts = rfm['Segment'].value_counts().reset_index()
segment_counts.columns = ['Segment', 'Klientide arv']

fig = px.bar(
    segment_counts,
    x='Segment',
    y='Klientide arv',
    title='UrbanStyle kliendisegmentide jaotus',
    labels={
        'Segment': 'Kliendisegment',
        'Klientide arv': 'Klientide arv'
    }
)

fig.show()
```

3. Diagramm 2 - viimane ost vs kogukulutus.

```python
fig = px.scatter(
    rfm,
    x='Recency',
    y='Monetary',
    color='Segment',
    size='Frequency',
    title='UrbanStyle kliendisegmendid (RFM)',
    labels={
        'Recency': 'Päevi viimasest ostust',
        'Monetary': 'Kogukulutus (EUR)',
        'Frequency': 'Ostude arv',
        'Segment': 'Kliendisegment'
    }
)

fig.show()
```

4. Diagramm 3 — TOP 10 VIP klienti kogukulutuse järgi.

```python
top_vip = (
    rfm[rfm['Segment'] == 'VIP Champions']
    .nlargest(10, 'Monetary')
)

fig = px.bar(
    top_vip,
    x=top_vip.index,
    y='Monetary',
    title='Top 10 VIP klienti kogukulu järgi',
    labels={
        'x': 'Kliendi ID',
        'Monetary': 'Kogukulu'
    }
)

fig.update_xaxes(title_text='Kliendi ID')

fig.show()
```

5. Leian VIP klientide osakaalu kogu RFM Monetary summast. Kontrollin VIP ja At Risk segmendi klientide arvu.

```python
vip = rfm[rfm['Segment'] == 'VIP Champions']

vip_share = vip['Monetary'].sum() / rfm['Monetary'].sum() * 100

print(f"VIP klientide käibeosakaal: {vip_share:.1f}%")


print(f"VIP klientide arv: {len(vip)}")


at_risk = rfm[rfm['Segment'] == 'At Risk']

print(f"At Risk klientide arv: {len(at_risk)}")
```
